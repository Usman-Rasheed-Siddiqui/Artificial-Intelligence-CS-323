```python
import pandas as pd
```

```python
data = pd.read_csv('Heart_Dataset.csv')
data
```

```python
data.isnull().sum()

# To drop null row:
data = data.dropna()

# To drop a column
data = data.drop(columns=['Oldpeak'])

from sklearn.impute import SimpleImputer

# Different null value fillers
imputer = SimpleImputer(strategy='mean')
imputer = SimpleImputer(strategy='median')
imputer = SimpleImputer(strategy='most_frequent')

data[['Age', 'RestingBP', 'Cholesterol', 'MaxHR', 'Oldpeak']] = imputer.fit_transform(
    data[['Age', 'RestingBP', 'Cholesterol', 'MaxHR', 'Oldpeak']]
)
```

```python
print(data['Age'].max())
print(data['Age'].min())
```

```python
data['Age'] = ((data['Age'] - data['Age'].min())/(data['Age'].max() - data['Age'].min()))
data
```

```python
print(data['Age'].max())
print(data['Age'].min())
```

```python
data = pd.get_dummies(data, columns=['Gender'])
data
```

```python
data['RestingBP'] = ((data['RestingBP'] - data['RestingBP'].min())/(data['RestingBP'].max() - data['RestingBP'].min()))
data['Cholesterol'] = ((data['Cholesterol'] - data['Cholesterol'].min())/(data['Cholesterol'].max() - data['Cholesterol'].min()))
data['MaxHR'] = ((data['MaxHR'] - data['MaxHR'].min())/(data['MaxHR'].max() - data['MaxHR'].min()))
data['Oldpeak'] = ((data['Oldpeak'] - data['Oldpeak'].min())/(data['Oldpeak'].max() - data['Oldpeak'].min()))
data
```

```python
print(data['ExerciseAngina'].value_counts())
print(data['RestingECG'].value_counts())
print(data['ChestPainType'].value_counts())
```

```python
data = pd.get_dummies(data, columns=['ExerciseAngina'])
data = pd.get_dummies(data, columns=['RestingECG'])
data = pd.get_dummies(data, columns=['ST_Slope'])
data
```

```python
data['ChestPainType'] = data['ChestPainType'].map({'ATA': 1, 'NAP': 2, 'TA': 3, 'ASY': 4 })
data
```

```python
data['ChestPainType'] = ((data['ChestPainType'] - data['ChestPainType'].min())/(data['ChestPainType'].max() - data['ChestPainType'].min()))
data
```
