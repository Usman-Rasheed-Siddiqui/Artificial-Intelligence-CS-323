```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import OrdinalEncoder
from sklearn.preprocessing import StandardScaler
```

```python
data = pd.read_csv('Heart_Dataset.csv')
data
```

```python
# Min Max Scaling
scalar = MinMaxScaler()
data[['Age', 'RestingBP', 'Cholesterol']] = pd.DataFrame(scalar.fit_transform(data[['Age', 'RestingBP', 'Cholesterol']]))
data = data.rename(columns={
    'Age': 'Age_normalized',
    'RestingBP':'RestingBP_normalized',
    'Cholestrol': 'Cholestrol_normalized'
})
data
```

```python
standardize = StandardScaler()
data[['MaxHR', 'Oldpeak']] = pd.DataFrame(standardize.fit_transform(data[['MaxHR', 'Oldpeak']]))
data = data.rename(columns={
    'MaxHR': 'MaxHR_standardized', 
    'Oldpeak': 'Oldpeak_standardized'
})
data
```

```python
print(data['Oldpeak_standardized'].min())
print(data['Oldpeak_standardized'].max())
```

```python
encoder = OneHotEncoder(sparse_output=False)

columns = ['Gender', 'FastingBS', 'ExerciseAngina', 'ST_Slope']
encoded = encoder.fit_transform(data[columns])
encoded_df = pd.DataFrame(
    encoded,
    columns=encoder.get_feature_names_out(columns),
    index=data.index
)

data = pd.concat(
    [data.drop(columns=columns), encoded_df],
    axis=1
)

data
```

```python
ordinal = OrdinalEncoder(
    categories=[['TA', 'ATA', 'NAP', 'ASY']]
)

data[['ChestPainType']] = pd.DataFrame(ordinal.fit_transform(data[['ChestPainType']]))
data
```

```python
data['RestingECG'] = data['RestingECG'].map({'Normal': 'low', 'LVH': 'medium', 'ST': 'high'})
data
```
